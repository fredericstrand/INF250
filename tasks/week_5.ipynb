```python
from skimage.morphology import square, disk, dilation, erosion, opening, closing
from skimage.filters import rank
from skimage import img_as_ubyte
import matplotlib.pyplot as plt
from skimage.io import imread
from skimage import io
```

# Task 1

```python
image = io.imread("images/rhino_d.tif")
plt.imshow(image, cmap='gray')
```

```python
dilated = dilation(image, square(3))
plt.imshow(dilated, cmap='gray')
```

```python
eroded = erosion(image, square(3))
plt.imshow(eroded, cmap='gray')
```

```python
dilate_after_erode = dilation(eroded, square(3))
plt.imshow(dilate_after_erode, cmap='gray')
```

```python
erode_then_dilate = dilation(eroded, square(3))
plt.imshow(erode_then_dilate, cmap='gray')
```

```python
opened = opening(image, square(3))
plt.imshow(opened, cmap='gray')
```

```python
closed = closing(image, square(3))
plt.imshow(closed, cmap='gray')
```

```python
image_ubyte = img_as_ubyte(image) * 255

min_r1 = rank.minimum(image_ubyte, disk(1))
min_r3 = rank.minimum(image_ubyte, disk(3))
max_r1 = rank.maximum(image_ubyte, disk(1))
max_r3 = rank.maximum(image_ubyte, disk(3))

fig, ax = plt.subplots(2, 2, figsize=(8, 6))
ax[0, 0].imshow(min_r1, cmap='gray'); ax[0, 0].set_title('Min (r=1)')
ax[0, 1].imshow(min_r3, cmap='gray'); ax[0, 1].set_title('Min (r=3)')
ax[1, 0].imshow(max_r1, cmap='gray'); ax[1, 0].set_title('Max (r=1)')
ax[1, 1].imshow(max_r3, cmap='gray'); ax[1, 1].set_title('Max (r=3)')
for a in ax.ravel(): a.axis('off')
plt.tight_layout()
plt.show()
```
